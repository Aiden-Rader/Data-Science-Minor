```python
"""
For example: x=5

You can use the print() function to view an object and check the variable type using the type() function
To view x, print(x)
To view the type value of x, print(type(x))
"""

# In Python, variables are created when you assign a value to them.

# Data Types

# Integers (int): Whole numbers without decimal points.
age = 20
print(age, type(age))

# Floats (float): Numbers with decimal points.
price = 99.09

print(price, type(price))

# Strings (str): Sequences of characters enclosed in single or double quotes.
name = "Alice"
greeting = 'Hello, world!'

print(name, type(name))

# Booleans (bool): Represent truth values (True or False).
is_adult = True

print(is_adult, type(is_adult))

# Type Conversion
# You can convert between different data types.

# Convert an integer to a string
number = 123
number_str = str(number)
print(number_str, type(number_str))

# Convert a string to an integer
string_number = "456"
int_number = int(string_number)
print(int_number, type(int_number))

# Convert an integer to a float
int_to_float = float(age) #age is an integer declared previously
print(int_to_float, type(int_to_float))

```

## **Data Structures**:

- **Tuples**

```python
# Lists: Ordered, mutable sequences of items.
my_list = [1, 2, 3, "apple", 4.5]  # Lists can contain different data types

# Indexing (accessing elements by position)
first_element = my_list[0]  # Accesses the first element (index 0)
last_element = my_list[-1]  # Accesses the last element
#print(f"First element: {first_element}, Last element: {last_element}")


# Slicing (extracting sub-lists)
sub_list = my_list[1:4]  # Elements from index 1 up to (but not including) index 4
#print(f"Sub-list: {sub_list}")

# Tuples: Ordered, immutable sequences. Like lists, but can't be modified after creation.
my_tuple = (10, 20, "banana", 30.5)

# Accessing elements (same as lists)
first_tuple_element = my_tuple[0]
# print(f"First element of tuple: {first_tuple_element}")


# Dictionaries: Key-value pairs. Use
my_dict = {"name": "Alice", "age": 30, "city": "New York"}

# Accessing values using keys
name = my_dict["name"]  # Accesses the value associated with the key "name"
#print(f"Name: {name}")


# Iterating through a dictionary
#print("\nDictionary elements:")
for key, value in my_dict.items():
	print(f"Key: {key}, Value: {value}")
```
