```python
import numpy as np
import matplotlib.pyplot as plt # matplotlib.pyplot is used for basic visualizations
import scipy.ndimage as ndimage # used only for rotation of 45 degrees since numpy does not allow us to do that... easily

# 1. Create a 5x5 NumPy array filled with random integers between 1 and 100.
rand_array = np.random.randint(1,100, size=(5, 5))

#    Then, find the minimum, maximum, and mean of the elements in the array.
min = np.min(rand_array)
max = np.max(rand_array)
print(f'The Array looks like: \n{rand_array}')
print(f'Minimum value: {min} \nMaximum value: {max} \n')

# 2. Create two 1D NumPy arrays of the same size with random integers.
array_one = np.random.randint(0, 100, size=10)
array_two = np.random.randint(0, 100, size=10)

#    Perform element-wise addition, subtraction, multiplication, and division.
add = array_one + array_two
sub = array_one - array_two
multi = array_one * array_two

# For dividing by zero we should use a try-except clause
try:
	div = array_one / array_two
except ZeroDivisionError:
	print("Error: Division by zero!")
	div = np.nan

#    Calculate the dot product of these two arrays.
dot_prod = np.dot(array_one, array_two)

print("Array One: ", array_one)
print("Array Two: ", array_two)
print("Addition: ", add)
print("Subtraction: ", sub)
print("Multiplication: ", multi)
print("Division: ", div)
print("Dot Product: ", dot_prod)
print('\n')

# 3. Create a 3x3 NumPy array. Rotate the array by 90 degrees clockwise and then by 45 degrees counterclockwise.
og_array = np.arange(9).reshape(3,3)
rotated_array_90 = np.rot90(og_array, -1)
rotated_array_45 = ndimage.rotate(rotated_array_90, 45, reshape=False)

#    Print the original and rotated arrays.
print(f'Original Array: \n{og_array}')
print(f'90 Degree Rotated Array: \n{rotated_array_90}')
print(f'45 Degree Rotated Array: \n{rotated_array_45}')
print('\n')

# 4. Create a 1D NumPy array with 10 elements. Reshape it to a 2x5 matrix.
another_array = np.arange(10)
reshaped_array = another_array.reshape(2,5)

#   Then, extract the element of the second row, third column.
print(f'This is the (2, 3) of the array: {reshaped_array[1,2]}')

# Create a meshgrid
x = np.arange(-30, 20, 5)
y = np.arange(-30, 20, 5)
x1, y1 = np.meshgrid(x, y)

# 5. Create a checkerboard(s) pattern with dimensions 10x10
checkerboard1 = np.zeros((x1.shape[0], x1.shape[1]))
checkerboard2 = np.zeros((x1.shape[0], x1.shape[1]))
checkerboard3 = np.zeros((x1.shape[0], x1.shape[1]))

# We use i for one axis and j for the other axis in the multi-dim array,
# range(10) is because it is a 10x10 multi dimensional array so we scan through i and j

# Option 1 (Using nested for loops with meshgrid)
for i in range(x1.shape[0]):
	for j in range(x1.shape[1]):
		if (i + j) % 2 == 0:
			checkerboard1[i, j] = 1  # Set value for even sum
		else:
			checkerboard1[i, j] = 0  # Set value for odd sum

# Option 2 (using np.where)
checkerboard2 = np.where((x1 + y1) % 2 == 0, 1, 0)  # Similar to a turnary operator

# Option 3 (you have the freedom to write it as you please)
# I am going to try and use indices (this is weird because its like the squares are flipped)
checkerboard3 = np.indices(x1.shape).sum(axis=0) % 2

# Display all three checkerboards
plt.figure(figsize=(12, 4))

# Option 1
plt.subplot(1, 3, 1)
plt.imshow(checkerboard1, cmap='gray', interpolation='nearest')
plt.title('Option 1: Nested Loops Using Meshgrid')
plt.axis('off')

# Option 2
plt.subplot(1, 3, 2)
plt.imshow(checkerboard2, cmap='gray', interpolation='nearest')
plt.title('Option 2: np.where')
plt.axis('off')

# Option 3
plt.subplot(1, 3, 3)
plt.imshow(checkerboard3, cmap='gray', interpolation='nearest')
plt.title('Option 3: np.indices')
plt.axis('off')

plt.tight_layout()
plt.show()
```
