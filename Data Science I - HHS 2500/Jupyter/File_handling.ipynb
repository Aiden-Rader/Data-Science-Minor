### **Relative Paths:**
Specify the location of a file relative to the current working directory of your program.

### **File Types**
Different files store data in different formats, each with its own structure and purpose. Examples:

  * Text files (.txt, .csv, .log): Store plain text data.
  
  * Binary files (.bin, .exe, .jpg, .mp3): Store non-text data, often in a format specific to an application.
  * Data files (.json, .xml, .csv): Store structured data in a format that's easy for programs to parse.
  * Other: `.jpg`, `.png`, `.gif`: Image files, `.mp3`, `.wav`: Audio files, `.mp4`, `.avi`: Video files, `.py`: Python source code files.


````python
"""
Modes of Operation: The mode determines how a file is opened:

'r' (read): Opens a file for reading.  (Default mode)
'w' (write): Opens a file for writing. Creates a new file if it doesn't exist; otherwise, truncates the existing file.
'a' (append): Opens a file for writing. Appends data to the end of the file if it exists, otherwise creates a new file.
'x' (exclusive creation): Creates a new file, but fails if the file already exists.
'b' (binary): Opens the file in binary mode.  Often combined with other modes (e.g., 'rb', 'wb').
't' (text): Opens the file in text mode (default).
"""

"""Demonstrates basic file operations."""
filepath = "C:/Users/aiden/OneDrive - University of Toledo/Fall 2024 UT/Data Science I/Jupyter Notebook/my_file.txt"
# 'w' mode: Create/overwrite the file
with open(filepath, 'w') as file:
	file.write("This is the first line.\n")
	file.write("This is the second line.\n")

# 'r' mode: Read the file
with open(filepath, 'r') as file:
	contents = file.read()
	print("File contents (read mode):\n", contents)

# 'a' mode: Append to the file
with open(filepath, 'a') as file:
	file.write("This line is appended.\n")

# 'r+' mode: Read and write
with open(filepath, 'r+') as file:
	file.write("This line is inserted at the beginning.\n")
	file.seek(0)  # Rewind to the beginning of the file
	contents = file.read()
	print("\nFile contents (r+ mode):\n", contents)


import os

print("Current Working Directory:", os.getcwd())  #Get current working directory
relative_path = "my_file.txt"
absolute_path = os.path.abspath(relative_path)  # Convert relative to absolute

print("Relative Path:", relative_path)
print("Absolute Path:", absolute_path)

os.path.exists(relative_path) # checks if the filename exists in the current working directory
with open(relative_path,'w') as f: #creating a file
	f.write("This is some text for the file.")
print("File created successfully.")
os.path.exists(relative_path) # checks if the filename exists in the current working directory

with open(relative_path) as f: #reading the file we created
	text = f.read()
	print(text)
	f.close()
````
