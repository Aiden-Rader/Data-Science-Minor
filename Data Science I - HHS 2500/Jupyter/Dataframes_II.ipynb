# **Dataframes II**
**Introduction to Pandas and DataFrames**

DataFrame is a two-dimensional data structure where data is aligned in a tabular fashion in rows and columns.
- Columns are represented by column names (or headers).
- Rows are indexed using an index number which starts from 0 by default.

Pandas is a powerful Python library for data analysis and manipulation.
DataFrames are its core data structure, offering a tabular way to represent data.

The main data structures *pandas* provides are *Series* and *DataFrames*. After a brief introduction to these two data structures and data ingestion, the key features of *pandas* this notebook covers are:
* Generating descriptive statistics on data
* Data cleaning using built in pandas functions
* Frequent data operations for subsetting, filtering, insertion, deletion and aggregation of data
* Merging multiple datasets using dataframes
* Working with timestamps and time-series data

````python
# Define file paths as a list of strings
file_paths = [
	'../ratings.csv',
	'../tags.csv',
	'../movies.csv'
]

# Read each file into its own DataFrame
df_ratings = pd.read_csv(file_paths[0])
df_tags = pd.read_csv(file_paths[1])
df_movies = pd.read_csv(file_paths[2])

# Display the first 10 rows of each DataFrame
print(f"First 10 rows of {file_paths[0]}:")
print(df_ratings.head(10), '\n')

print(f"First 10 rows of {file_paths[1]}:")
print(df_tags.head(10), '\n')

print(f"First 10 rows of {file_paths[2]}:")
print(df_movies.head(10), '\n')
````

````python
# 3. Get the most occuring rating
most_occuring_rating = df_ratings['rating'].mode()

# Display the result
print("Most occurring rating(s):")
print(most_occuring_rating)
````

````python
# 4. Using the .any() command, are there any ratings > 4?
is_greater_than_4 = (df_ratings['rating'] > 4).any()
print('\n')
print('Is there any rating greater than 4? ')
print(is_greater_than_4)
````

````python
# 1. Load the melb_data.csv and explore some descriptive statistics
df_melb_data = pd.read_csv('../melb_data.csv')
print(df_melb_data[:10])
print('\n')
df_melb_data.describe()
df_melb_data.head()
````

````python
# 2. Use .corr() to explore the dataset for columns that would be useful for predicting housing prices

# Select only numerical columns
numerical_data = df_melb_data.select_dtypes(include=['number'])

# Calculate correlation matrix
correlation_matrix = numerical_data.corr()

# Display correlations with the "Price" column
price_correlations = correlation_matrix['Price'].sort_values(ascending=False)
print("Correlation of numerical features with Price:")
print(price_correlations)

````

````python
# 3. Choose a method of interpolation for the missing values, and explain the reason for your choice

# Select numerical columns for interpolation
numerical_data = df_melb_data.select_dtypes(include=['number'])

# Apply linear interpolation for numerical columns
df_melb_data[numerical_data.columns] = numerical_data.interpolate(method='linear')

# Interpolate the 'YearBuilt' column separately (if required, though it is numerical but its kinda weird)
df_melb_data['YearBuilt'] = df_melb_data['YearBuilt'].interpolate(method='linear')

# Fill with a specific value or unknown in this case
# (can also use forward or backfill not sure which one is better though)
df_melb_data['CouncilArea'] = df_melb_data['CouncilArea'].fillna('Unknown')

# Display summary of missing values after interpolation
print("Missing values after interpolation:")
print(df_melb_data.isnull().sum())
````

````python
# 4. Filter for houses that cost < 400,00 AND sellerG contains the letters 'a' OR 'A' AND 'L' OR 'l'.
filtered_houses = df_melb_data[
	(df_melb_data['Price'] < 400000) &
	(df_melb_data['SellerG'].str.contains('a', case=False)) &
	(df_melb_data['SellerG'].str.contains('l', case=False))
]

# Display the filtered DataFrame
print(filtered_houses.head())
````

````python
import pandas as pd
import numpy as np

# Sample DataFrame (replace with your actual data)
data = {'Category': ['A', 'A', 'B', 'B', 'C', 'C'],
        'Value1': [10, 15, 20, 25, 30, 35],
        'Value2': [5, 10, 15, 20, 25, 30]}
df = pd.DataFrame(data)

# 1: Group by 'Category' and calculate the sum of 'Value1' and 'Value2'
grouped_sum = df.groupby('Category').sum()
print(grouped_sum)

# 2: Group by 'Category' and calculate the mean of 'Value1' and the sum of 'Value2'
grouped_agg = df.groupby('Category').agg({'Value1': 'mean', 'Value2': 'sum'})
print(grouped_agg)

# 3: Group by 'Category' and calculate the count of items in each category
grouped_count = df.groupby('Category').count()
print(grouped_count)
````
